## Project Title:

Amyloid, Tau, and Brain Weight in Alzheimer's Dementia: Differences by Cognitive Status and Sex



## Project Goal:

This project seeks to determine whether brain levels of amyloid-beta (ABeta40 and ABeta42) and phosphorylated tau (pTAU), and brain weight, differ between brain donors with dementia and donors without dementia, and whether sex modifies these differences. We will use Student's t-tests to compare donors with and without dementia, and a one-way ANOVA followed by Tukey's post hoc test to compare four groups (healthy female, healthy male, dementia female, dementia male). We hypothesize that donors with dementia will have higher amyloid and tau levels and lower brain weight than donors without dementia. To connect the molecular measures to cognition, we will also use linear regression to test whether pTAU levels predict a donor's last MMSE score. Because these tissue measurements are highly variable and the sample is small (84 donors), we will also examine how skewed data and limited group sizes affect what conclusions we can draw.

## Disease Background: 

* Prevalence & incidence

Alzheimer's disease is prevalent in an estimated 7.2 million Americans ages 65 and older (around 1 in 9 older adults). This number is projected to reach roughly 13.8 million by the year 2060. Even though the absolute number of people developing Alzheimer's is rising due to the aging population, age specific incidence and prevalence rates in the US and other high-income countries have experienced moderate declines, largely due to rising education levels. That being said, findings remain mixed, as a review demonstrated overall dementia incidence declining worldwide from 1977 to 2017, the incidence of Alzheimer's dementia specifically remained steady, and these encouraging reductions may not extend to all races and ethnic groups or to low and middle income countries.

Source: [National Library of Medicine: 2025 Alzheimer's disease facts and figures](https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/)

* Economic burden

Alzheimer's disease has a significant economic impact on public healthcare programs, long-term care systems, and families. Total payments in 2025 for health-care and hospice services for individuals ages 65 and older with dementia reach ~$384 billion. The average per-person payments for beneficiaries ages 65 and older with dementia are nearly 3x higher than those without cognitive impairment. Medicaid payments are also 22x greater due to long-term nursing and institutional care demands. In 2024, nearly 12 million family members and other unpaid caregivers provided around 19.2 billion hours of care.

Source: [National Library of Medicine: 2025 Alzheimer's disease facts and figures](https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/)

* Risk factors (genetic, lifestyle) 

When breaking down Alzheimer’s risk, age is the primary factor. Most diagnoses hit at 65 or older, jumping from about one in 13 people between 65 and 84 to one in three once you pass 85. Genetics and family history also drive a huge chunk of that baseline risk. Having a parent or sibling with the disease increases your odds, largely influenced by inherited variants in the APOE gene (where carrying the ε4 allele elevates susceptibility, while ε2 appears somewhat protective). Lifestyle and vascular health are heavily intertwined with neurodegeneration. Factors that compromise cardiovascular function, like chronic hypertension, high LDL cholesterol, diabetes, obesity, smoking, and a sedentary lifestyle, significantly amplify risk. Moreover, habits like poor sleep, alcohol misuse, lack of cognitive engagement, and social isolation, along with clinical issues like unaddressed hearing loss, depression, and a history of concussions or traumatic brain injuries.


Source: [Thinking About Your Risk for Alzheimer’s Disease? Five Questions To Consider](https://www.nia.nih.gov/health/alzheimers-causes-and-risk-factors/thinking-about-your-risk-alzheimers-disease-five)

* Societal determinants

Social determinants of health play a massive role in driving risk and cognitive decline. Lower socioeconomic status and limited education consistently correlate with a higher incidence of ADRDs, whereas higher educational attainment and financial stability effectively slow down time to diagnosis. Work environment and resource availability are also major factors as individuals engaged in manual labor face higher ADRD rates, as do those experiencing food insecurity across early or late life, or living in economically disadvantaged neighborhoods stripped of core physical infrastructure. Metabolic shifts like high midlife BMI followed by late-life wasting, alongside chronic elevated stress and early-life adversity, further accelerate neurodegenerative vulnerability, while consistent social engagement shows a measurable protective effect. Racial discrimination also drives up ADRD rates both directly through physiological stress pathways and indirectly by enforcing disparities in education, employment options, and residential segregation. Ultimately, tackling Alzheimer's risk isn't just about clinical interventions. It also requires addressing root systemic factors like education policy, socioeconomic equity, and structural discrimination.

Source: [Effect of Social Determinants of Health on Cognition and Risk of Alzheimer Disease and Related Dementias
](https://pubmed.ncbi.nlm.nih.gov/34103175/)

* Symptoms

Memory loss is the primary symptom of Alzheimer’s disease, typically kicking off with someone forgetting recent conversations, events, repeating questions, and misplacing items. As neurodegeneration spreads, the deficits hit abstract thinking and executive functioning which makes everyday logic, multitasking, and other issues difficult. Judgment and decision-making erode, resulting in inappropriate social choices, wearing weather-inappropriate clothes, or failing to handle basic emergencies like burnt food or driving decisions. Everyday routines, like following a recipe or playing a game, become increasingly harder to do, eventually advancing to the point where individuals lose the ability to handle fundamental daily tasks like dressing and bathing. Alongside cognitive decline, structural brain changes alter mood and personality, frequently triggering depression, social withdrawal, severe mood swings, paranoia, aggression, sleep disruptions, wandering, and delusions. Interestingly, even deep into disease progression, preserved procedural and emotional skills often linger. This includes singing, listening to music, dancing, or any other creative art. The brain regions governing those pathways are spared until the terminal stages, where widespread functional loss eventually disrupts mobility, swallowing, and basic physiologic autonomy.

Source: [Mayoclinic: Alzheimer's Disease
](https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/symptoms-causes/syc-20350447)

* Diagnosis

To diagnose Alzheimer's disease, clinicians begin by asking about the symptoms someone is feeling, family history, and neurological exams to establish functional decline. They use standard blood work to help rule out any thyroid dysfunctions or vitamin deficiencies. Neuropsychological testing then benchmarks cognitive deficits against demographic cohorts to track progression. For structural evaluation, CT scans rule out acute lesions, while MRI is favored to identify regional brain atrophy and provide baseline safety imaging for monoclonal antibody treatments. Finally, biological confirmation relies on advanced biomarkers. FDG-PET highlights diagnostic hypometabolic patterns, amyloid and tau PET map protein burden, and CSF or emerging plasma assays quantify beta-amyloid and phosphorylated tau levels. Additionally, APOE ε4 genotyping is increasingly used to screen for brain swelling and microhemorrhage risks before initiating anti-amyloid therapies like lecanemab or donanemab.

Source: [Mayoclinic: Alzheimer's Disease - Diagnosis
](https://www.mayoclinic.org/diseases-conditions/alzheimers-disease/diagnosis-treatment/drc-20350453)

* Standard of care treatments (& reimbursement) 

Two FDA-approved antibody drugs, lecanemab (Leqembi) and donanemab (Kisunla), slow progression by removing amyloid plaques, and both are given by IV infusion. Medicare requires patient data to be submitted to a clinical registry for coverage of either drug. The older symptom-relief drugs are the cholinesterase inhibitors and memantine. Both antibodies require confirmed brain amyloid, and both can cause ARIA (brain swelling or bleeding), with the highest risk in people who carry two APOE4 copies.

Source: [Baylor Medicine Treatment for Alzheimer's Disease and other Dementias](https://www.bcm.edu/healthcare/specialties/neurology/alzheimers-disease-and-memory-disorders/services/alzheimers-disease-treatment)

* Disease progression & prognosis

Alzheimer's follows a long, progressive course that begins years before symptoms, moving from preclinical disease to mild cognitive impairment (MCI) to dementia. Risk of progression rises with each stage: in one meta-analysis the share of people progressing from preclinical AD to MCI was 20% at stage 1, 38% at stage 2 and 73% at stage 3. Data on survival are limited, but median survival with AD dementia has been estimated at about 7 years from the onset of cognitive decline. Factors linked to faster progression include APOE4 status, entorhinal cortex atrophy, elevated CSF tau, and lower MMSE scores.

Source: [NIH Alzheimer’s Disease: Epidemiology and Clinical Progression](https://pmc.ncbi.nlm.nih.gov/articles/PMC9095793/#Sec1)

* Continuum of care providers

Dementia care requires collaboration among family members, medical professionals, other health service providers, and communities. The professionals involved include primary care doctors, neurologists, geriatricians, geriatric psychiatrists, neuropsychologists, social workers and psychologists, speech, physical and occupational therapists, nurses, and home health aides. They may see patients at a clinic, hospital, academic hospital, or long-term care facility.

Source: [Alzheimer's.gov Resources for Health Care Providers: Alzheimer’s and Related Dementias](https://www.alzheimers.gov/professionals/health-care-providers)

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)

Anatomy and organ level: Memory depends on the hippocampus and its associated structures, including the entorhinal cortex. Early Alzheimer's changes are concentrated there, with entorhinal atrophy and tangles, loss of layer II neurons that disrupts the pathway into the hippocampus, and tangle deposition in the hippocampal CA1 region. Up to 80% of hippocampal neurons may die over the course of the disease, and late-stage disease brings widespread neuron loss and cortical atrophy.

Cell and molecular level: The hallmarks are extracellular amyloid-beta plaques and intracellular neurofibrillary tangles. Amyloid-beta is cut from the amyloid precursor protein (APP), and the Aβ42 form is thought to be most involved in disease. Tangles form when tau becomes hyperphosphorylated and detaches from microtubules, which disrupts transport inside neurons and can lead to cell death. Synapse loss correlates strongly with cognitive decline. Chronic overactivation of NMDA glutamate receptors (excitotoxicity) is another proposed mechanism.

Sources: [NIH What Happens to the Brain in Alzheimer's Disease?](https://www.nia.nih.gov/health/alzheimers-causes-and-risk-factors/what-happens-brain-alzheimers-disease) [NIH Mechanisms of Alzheimer’s Disease Pathogenesis and Prevention: The Brain, Neural Pathology, N-methyl-D-aspartate Receptors, Tau Protein and Other Risk Factors](https://pmc.ncbi.nlm.nih.gov/articles/PMC5290713/)

* Clinical Trials/next-gen therapies

As of 2024, ClinicalTrials.gov listed 187 trials targeting Alzheimer's disease and MCI due to AD, with 36 drugs in phase III, 87 in phase II and 31 in phase I. Most trials still focus on amyloid and tau, and the main trends are disease-modifying therapies and enrolling people at earlier stages, including MCI and cognitively healthy people with biomarker evidence of disease. Several late-stage anti-amyloid approaches have failed, including gamma-secretase inhibitors, BACE1 inhibitors, solanezumab and gantenerumab. Lecanemab received full FDA approval in 2023 based on the Clarity AD phase 3 trial, and donanemab had completed phase 3 and was awaiting approval when these reviews were written. Both drugs carry a risk of amyloid-related imaging abnormalities (ARIA), and the reviews note high monitoring and treatment costs and a limited range of indications. Next-generation approaches include anti-tau antibodies (bepranemab, E2814), an antisense drug that lowers tau production (BIIB080), anti-inflammatory and neuroprotective drugs, and newer drug designs such as dual-target inhibitors and PROTACs.

Sources: [Nature: Recent advances in Alzheimer’s disease: mechanisms, clinical trials and new drug development strategies](https://www.nature.com/articles/s41392-024-01911-3) [NIH Clinical trials of new drugs for Alzheimer disease: a 2020–2023 update](https://pmc.ncbi.nlm.nih.gov/articles/PMC10544555/)

## Data-Set:

**Source.** We used donor data from the **Seattle Alzheimer's Disease Brain Cell Atlas (SEA-AD)**, given to us as `Metadata_and_Protein_Data_for_Module_1.csv` (84 donors, one row per donor). SEA-AD is a public resource created by the Allen Institute for Brain Science and the University of Washington. The donors came from two long-term research studies: Adult Changes in Thought (ACT, 69 donors) and the UW Alzheimer's Disease Research Center (ADRC, 15 donors). The data are available at [brain-map.org/consortia/sea-ad](https://brain-map.org/consortia/sea-ad/our-data)

**How the data were collected.**
* **Donors:** People were tested for memory and thinking during life (for example the MMSE, scored 0-30) and then donated their brains at death. Half of the donors had dementia (42) and half did not (42). They were 65-102 years old at death, and 51 were female and 33 were male. Each donor's last MMSE was taken a median of about 28 months before death.
* **Brain tissue:** Brains were collected soon after death (3-11 hours) and frozen. `Fresh Brain Weight` is the weight of the brain at autopsy, in grams.
* **Protein levels:** ABeta40, ABeta42, and phosphorylated tau (pTAU) were measured in the middle temporal gyrus, a brain region affected early in Alzheimer's disease, using a Luminex immunoassay. Units are picograms of protein per microgram (pg/ug) of total protein.

**Variables we used.**

| Variable | Column in the CSV | Unit |
|---|---|---|
| Cognitive status | `Cognitive Status` ("Dementia" / "No dementia") | category |
| Sex | `Sex` ("Female" / "Male") | category |
| Amyloid and tau | `ABeta40 pg/ug`, `ABeta42 pg/ug`, `pTAU pg/ug` | pg/ug |
| Brain weight | `Fresh Brain Weight` | grams |
| Cognition score | `Last MMSE Score` | 0-30 |

**Data problems and how we handled them.**
1. **One brain weight is missing** (recorded as the text "Unavailable"). It becomes NaN, and that donor is left out of brain weight analyses only (n = 83).
2. **Four donors have no MMSE score**, so they are left out of the regression.
3. **One MMSE score is impossible:** a score of 33, when the maximum is 30. We treat it as a data-entry error and leave that donor out of the regression (final n = 79).
4. **The amyloid values are very spread out.** A few donors have extremely high values (ABeta40 ranges from about 0.0001 to 981 pg/ug), which makes t-tests less reliable. We discuss this in the limitations.
5. **The groups are uneven by sex.** The dementia group has 27 females and 15 males, and the no dementia group has 24 females and 18 males. This matters for brain weight, because males have heavier brains on average.

## Data Analysis: 

Our data has 84 brain donors, and each row is one donor. We turn each row into a Patient object so we can filter donors by sex and cognitive status. Missing values, like the brain weight marked "Unavailable", become NaN so they don't crash the code. We keep only the variables we need: sex, cognitive status, last MMSE score, brain weight, and the ABeta40, ABeta42, and pTAU levels. The protein levels are measured in pg per µg of total protein in the middle temporal gyrus, a brain region affected early in Alzheimer's disease.

```python
import csv
import numpy as np
import matplotlib.pyplot as plt
import statistics
from scipy import stats
from sklearn.linear_model import LinearRegression


# Turns a CSV cell into a number. Blank cells or text like "Unavailable" become NaN (not a number)
def to_number(text):
    try:
        return float(text)
    except ValueError:
        return float("nan")


class Patient:
    all_patients = []

    def __init__(self, donor_id: str, sex: str, cog_stat: str, mmse: float, brain_weight: float,
                 abeta40: float, abeta42: float, ptau: float):
        self.donor_id = donor_id
        self.sex = sex                     # "Female" or "Male"
        self.cog_stat = cog_stat           # "Dementia" or "No dementia"
        self.mmse = mmse                   # last MMSE score (0-30)
        self.brain_weight = brain_weight   # grams
        self.abeta40 = abeta40             # pg/ug
        self.abeta42 = abeta42             # pg/ug
        self.ptau = ptau                   # pg/ug
        Patient.all_patients.append(self)

    def __repr__(self):
        return (f"{self.donor_id}: ({self.sex} | {self.cog_stat} | MMSE {self.mmse} | "
                f"brain weight {self.brain_weight} | ABeta42 {self.abeta42} | pTAU {self.ptau})")

    @classmethod
    def instantiate_from_csv(cls, filename: str):
        with open(filename, encoding="utf-8-sig") as f:
            reader = csv.DictReader(f)
            rows_of_patients = list(reader)
            for row in rows_of_patients:
                Patient(
                    donor_id=row["Donor ID"],
                    sex=row["Sex"],
                    cog_stat=row["Cognitive Status"],
                    mmse=to_number(row["Last MMSE Score"]),
                    brain_weight=to_number(row["Fresh Brain Weight"]),
                    abeta40=to_number(row["ABeta40 pg/ug"]),
                    abeta42=to_number(row["ABeta42 pg/ug"]),
                    ptau=to_number(row["pTAU pg/ug"])
                )

    @classmethod
    def filter(cls, patients, sex: str = "any", cog_stat: str = "any"):
        all_patients = patients
        remove_list = []
        attr_list = (
            sex,
            cog_stat
        )
        attr_name = (
            "sex",
            "cog_stat"
        )
        for attr in range(len(attr_list)):
            if attr_list[attr] != "any":
                for patient in all_patients:
                    if getattr(patient, attr_name[attr]) != attr_list[attr]:
                        remove_list.append(patient)
                all_patients = [patient for patient in all_patients if patient not in remove_list]
                remove_list.clear()

        return all_patients


# Create one Patient object for every row in the CSV
Patient.instantiate_from_csv("DataSet/Metadata and Protein Data for Module 1.csv")

print(Patient.all_patients[0])
print(f'Number of patients = {len(Patient.all_patients)}')
print(f'Healthy females = {len(Patient.filter(Patient.all_patients, sex="Female", cog_stat="No dementia"))}')
print(f'Healthy males = {len(Patient.filter(Patient.all_patients, sex="Male", cog_stat="No dementia"))}')
print(f'Dementia females = {len(Patient.filter(Patient.all_patients, sex="Female", cog_stat="Dementia"))}')
print(f'Dementia males = {len(Patient.filter(Patient.all_patients, sex="Male", cog_stat="Dementia"))}')
```

We loaded 84 donors in four groups: 24 healthy females, 18 healthy males, 27 dementia females, and 15 dementia males. The two dementia groups add up to 42 donors, and so do the two healthy groups, so the dementia vs no dementia comparison is balanced. There are more females than males in both, and the dementia group has proportionally more females (64% vs 57%). That matters later for brain weight. The small size of the male groups (15 and 18) also means we have less power to detect differences in males. The printed first donor shows the objects were created correctly.

We ask: do donors with dementia have different levels of ABeta40, ABeta42, pTAU, or brain weight than donors without dementia? For each measure we find the mean and standard deviation of each group, run a Student's t-test, and draw a bar graph. Our cutoff is p < 0.05. A p-value below 0.05 means the difference is unlikely to be just chance. We chose these measures because amyloid-beta plaques and tau tangles are the two hallmarks of Alzheimer's disease, and brain loss should lower brain weight. 

Hypothesis: dementia donors have higher amyloid and tau, and lower brain weight. 

Each bar shows the group mean and the line shows one standard deviation. We ran four t-tests, so one significant result could still be a false positive.

```python
# ---------------------------------------------------------------
# STUDENT'S T-TEST + BAR GRAPHS: dementia vs. no dementia
# ---------------------------------------------------------------
# attribute name in the Patient class : label for the graph
measures = {
    "abeta40": "ABeta40 (pg/ug)",
    "abeta42": "ABeta42 (pg/ug)",
    "ptau": "pTAU (pg/ug)",
    "brain_weight": "Brain weight (g)"
}

plt.figure(figsize=(16, 4.5))
plot_number = 1

for attr in measures:
    # Collect the values for each group (skip any missing values)
    dementia_vals = []
    healthy_vals = []

    for patient in Patient.filter(Patient.all_patients, cog_stat="Dementia"):
        value = getattr(patient, attr)
        if not np.isnan(value):
            dementia_vals.append(value)

    for patient in Patient.filter(Patient.all_patients, cog_stat="No dementia"):
        value = getattr(patient, attr)
        if not np.isnan(value):
            healthy_vals.append(value)

    # Mean and standard deviation of each group
    x_healthy_bar = statistics.mean(healthy_vals)
    x_dementia_bar = statistics.mean(dementia_vals)
    healthy_stdev = statistics.stdev(healthy_vals)
    dementia_stdev = statistics.stdev(dementia_vals)

    print(f'{measures[attr]}')
    print(f'  No dementia: x_bar = {x_healthy_bar}, stdev = {healthy_stdev}, n = {len(healthy_vals)}')
    print(f'  Dementia:    x_bar = {x_dementia_bar}, stdev = {dementia_stdev}, n = {len(dementia_vals)}')

    # Student's t-test
    t_stat, p_val = stats.ttest_ind(healthy_vals, dementia_vals)
    print(f'  t_stat = {t_stat}, p_val = {p_val}')

    # Bar graph
    cols = ['No Dementia', 'Dementia']
    means = [x_healthy_bar, x_dementia_bar]
    stdevs = [healthy_stdev, dementia_stdev]
    yerr = [np.zeros(len(means)), stdevs]

    plt.subplot(1, 4, plot_number)
    plt.bar(cols, means, yerr=yerr, capsize=10, color=["blue", "orange"])
    plt.title(measures[attr])
    plt.xlabel("Cognitive Status")
    plt.ylabel(measures[attr])
    y_max = max(means) + max(stdevs) * 1.1
    plt.ylim(0, y_max * 1.25)
    plt.text(0.5, y_max, f"t = {t_stat:.2f}\np = {p_val:.3f}", ha='center', va='bottom')

    plot_number += 1

plt.tight_layout()
plt.show()
```

Amyloid and tau: the dementia bars are higher, as we predicted, but the differences are not statistically significant. The error bars are huge. For example, the ABeta42 standard deviation in the dementia group (218) is more than double its mean (97). A few donors with extremely high values make the data very spread out, so the test can't be sure the difference is real. The averages differ a lot (ABeta40 is about 5.7 times higher in the dementia group), so this may be a real difference that our small sample and skewed data can't confirm.

Brain weight: dementia donors' brains weigh about 58 g less (roughly 5%), and this is significant (p = 0.025). This matches our hypothesis, and the spread is small compared with the amyloid data (SD of about 115 g in both groups). But ANOVA shows sex plays a big part.

Previous test ignored sex, and males and females may differ. So we split donors into four groups (healthy female, healthy male, dementia female, dementia male) and use a one-way ANOVA to ask whether any of the four differ. If the ANOVA is significant (p < 0.05), we run Tukey's test to find which pairs differ. If it isn't, we stop, because there's nothing to follow up. We use Tukey's test instead of running many t-tests because comparing four groups gives six pairs, and Tukey's test adjusts for the extra chances of a false positive. This lets us ask whether the effect of dementia is different in males and females.

```python
# ---------------------------------------------------------------
# ONE-WAY ANOVA + TUKEY'S POST HOC TEST: healthy/dementia x female/male
# ---------------------------------------------------------------
# attribute name in the Patient class : label for the graph
measures = {
    "abeta40": "ABeta40 (pg/ug)",
    "abeta42": "ABeta42 (pg/ug)",
    "ptau": "pTAU (pg/ug)",
    "brain_weight": "Brain weight (g)"
}

plt.figure(figsize=(18, 5))
plot_number = 1

for attr in measures:
    # Collect the values for each of the 4 groups (skip any missing values)
    health_fem_vals = []
    health_male_vals = []
    diseased_fem_vals = []
    diseased_male_vals = []

    for patient in Patient.filter(Patient.all_patients, sex="Female", cog_stat="No dementia"):
        if not np.isnan(getattr(patient, attr)):
            health_fem_vals.append(getattr(patient, attr))
    for patient in Patient.filter(Patient.all_patients, sex="Male", cog_stat="No dementia"):
        if not np.isnan(getattr(patient, attr)):
            health_male_vals.append(getattr(patient, attr))
    for patient in Patient.filter(Patient.all_patients, sex="Female", cog_stat="Dementia"):
        if not np.isnan(getattr(patient, attr)):
            diseased_fem_vals.append(getattr(patient, attr))
    for patient in Patient.filter(Patient.all_patients, sex="Male", cog_stat="Dementia"):
        if not np.isnan(getattr(patient, attr)):
            diseased_male_vals.append(getattr(patient, attr))

    # Mean and standard deviation of each group
    x_health_fem_bar = statistics.mean(health_fem_vals)
    x_health_male_bar = statistics.mean(health_male_vals)
    x_diseased_fem_bar = statistics.mean(diseased_fem_vals)
    x_diseased_male_bar = statistics.mean(diseased_male_vals)

    health_fem_stdev = statistics.stdev(health_fem_vals)
    health_male_stdev = statistics.stdev(health_male_vals)
    diseased_fem_stdev = statistics.stdev(diseased_fem_vals)
    diseased_male_stdev = statistics.stdev(diseased_male_vals)

    print(f'{measures[attr]}')
    print(f'  Healthy Female:  x_bar = {x_health_fem_bar}, stdev = {health_fem_stdev}, n = {len(health_fem_vals)}')
    print(f'  Healthy Male:    x_bar = {x_health_male_bar}, stdev = {health_male_stdev}, n = {len(health_male_vals)}')
    print(f'  Diseased Female: x_bar = {x_diseased_fem_bar}, stdev = {diseased_fem_stdev}, n = {len(diseased_fem_vals)}')
    print(f'  Diseased Male:   x_bar = {x_diseased_male_bar}, stdev = {diseased_male_stdev}, n = {len(diseased_male_vals)}')

    # Run one-way ANOVA
    f_stat, p_value = stats.f_oneway(health_fem_vals, health_male_vals, diseased_fem_vals, diseased_male_vals)
    print(f'  F-statistic = {f_stat}, p-value = {p_value}')

    # Tukey's post hoc test: only interpret it if the ANOVA is significant (p < 0.05)
    # Group numbers in the table: 0 = Healthy Female, 1 = Healthy Male, 2 = Diseased Female, 3 = Diseased Male
    if p_value < 0.05:
        tukey_result = stats.tukey_hsd(health_fem_vals, health_male_vals, diseased_fem_vals, diseased_male_vals)
        print(tukey_result)
    else:
        print('  ANOVA not significant, so no post hoc test')

    # Bar graph
    sex_cols = ['Healthy Female', 'Healthy Male', 'Diseased Female', 'Diseased Male']
    means = [x_health_fem_bar, x_health_male_bar, x_diseased_fem_bar, x_diseased_male_bar]
    stdevs = [health_fem_stdev, health_male_stdev, diseased_fem_stdev, diseased_male_stdev]
    yerr = [np.zeros(len(means)), stdevs]

    plt.subplot(1, 4, plot_number)
    plt.bar(sex_cols, means, yerr=yerr, capsize=10, color=["red", "blue", "pink", "skyblue"])
    plt.title(measures[attr])
    plt.xlabel("Sex and Health Status")
    plt.ylabel(measures[attr])
    plt.xticks(rotation=30)
    y_max = max(means) + max(stdevs) * 1.1
    plt.ylim(0, y_max * 1.25)
    plt.text(1.5, y_max, f"One-Way-ANOVA: p = {p_value:.3g}", ha='center', va='bottom', fontsize=10)

    plot_number += 1

plt.tight_layout()
plt.show()
```

ABeta40 (p = 0.23), ABeta42 (p = 0.25), pTAU (p = 0.29): none of the ANOVAs are significant, so we conclude the four groups do not differ on amyloid or tau, and Tukey's test isn't run. This agrees with previous code, where none of these were significant either. Splitting into four groups also made each group smaller (as few as 15 donors), which lowers our ability to detect differences.

Brain weight (p ≈ 0.0000006): highly significant, so Tukey's test ran. Groups are numbered 0 = Healthy Female, 1 = Healthy Male, 2 = Dementia Female, 3 = Dementia Male. The significant pairs are:

Healthy Female vs Healthy Male
Healthy Male vs Dementia Female
Dementia Female vs Dementia Male

Key finding: healthy females vs dementia females (p = 0.42) and healthy males vs dementia males (p = 0.29) are not significantly different. Every significant difference is between males and females, and males have brains about 130 g heavier. So the lower brain weight in t-test is mostly because the dementia group has more females, not proof that dementia shrinks the brain in this sample. Within each sex, the dementia mean is still a bit lower than the healthy mean, so there may be a small real effect that we don't have enough donors to confirm.

Now we ask whether the brain measures predict how well a donor did on their last cognitive test (the MMSE, scored 0 to 30, where lower is worse). This links the molecular measurements to actual cognition. For each measure we draw a scatter plot and fit a straight line (linear regression). Three numbers matter:

Slope: how much the MMSE changes for each one-unit increase in the measure.
R²: how much of the variation in MMSE the line explains (0 = none, 1 = all).
p-value: whether the slope is really different from zero.

We left out one impossible MMSE of 33 (the maximum is 30) and donors with missing values, which leaves 79 donors (78 for brain weight). One thing to keep in mind is that the MMSE was often taken well before death (a median of about 28 months), so it may not match the brain tissue exactly.

```python
# ---------------------------------------------------------------
# SCATTER PLOTS + LINEAR REGRESSION: brain measures vs. last MMSE score
# ---------------------------------------------------------------
# attribute name in the Patient class : label for the graph
predictors = {
    "ptau": "pTAU (pg/ug)",
    "abeta42": "ABeta42 (pg/ug)",
    "brain_weight": "Brain weight (g)"
}

plt.figure(figsize=(16, 4.8))
plot_number = 1

for attr in predictors:
    # Collect the x values (brain measure) and y values (MMSE score) for each patient
    # Skip patients with a missing value, and skip the impossible MMSE score of 33 (the MMSE maximum is 30)
    brain_measure = []
    mmse_score = []

    for patient in Patient.all_patients:
        value = getattr(patient, attr)
        if not np.isnan(value) and not np.isnan(patient.mmse) and patient.mmse <= 30:
            brain_measure.append(value)
            mmse_score.append(patient.mmse)

    X = np.array(brain_measure).reshape(-1, 1)   # Independent variable, shape: (n_samples, 1)
    y = np.array(mmse_score)                     # Dependent variable, shape: (n_samples,)

    # Do the linear regression
    model = LinearRegression()
    model.fit(X, y)

    slope = model.coef_[0]
    intercept = model.intercept_
    r2 = model.score(X, y)

    # p-value for the slope (tests whether the slope is different from 0)
    p_value = stats.linregress(brain_measure, mmse_score).pvalue

    print(f'{predictors[attr]}: n = {len(X)}, slope = {slope}, intercept = {intercept}, R^2 = {r2}, p = {p_value}')

    # Annotate equation
    equation = f"y = {slope:.3f}x + {intercept:.2f}\nR² = {r2:.3f}, p = {p_value:.3g}"

    plt.subplot(1, 3, plot_number)
    plt.scatter(X, y, color='blue')
    plt.plot(X, model.predict(X), color="red")
    plt.ylim(0, 37)
    plt.text(X.min(), 31.5, equation, color="red", fontsize=11, verticalalignment='bottom')
    plt.xlabel(predictors[attr])
    plt.ylabel('Last MMSE Score')
    plt.title(f'Scatter Plot of {predictors[attr].split(" (")[0]} vs Last MMSE Score')

    plot_number += 1

plt.tight_layout()
plt.show()
```

pTAU: higher pTAU goes with lower MMSE (about half a point lower per 1 pg/ug), and this is significant. But R² is only 0.086, so pTAU explains only about 9% of the differences in cognition. It's a real but weak relationship, which fits the idea that many factors affect cognition. Interestingly, pTAU predicts MMSE even though it did not differ between the dementia and no dementia groups in student t-test.

ABeta42: no relationship (p = 0.27). The line is nearly flat, and the fit is pulled around by a few donors with extreme values.

Brain weight: heavier brains go with higher MMSE (about 1.4 points per 100 g), and it's the strongest of the three (R² = 0.135, so about 13.5% explained). Since males have heavier brains, part of this is a sex effect.

Important: these show association, not cause. We can't say that high pTAU causes lower scores.

## Conclusions and Ethical Implications:

**What we found.** Our hypothesis was only partly supported. Amyloid-beta and pTAU were higher on average in dementia donors, but the differences were not statistically significant (p = 0.080, 0.085, and 0.111) because the groups overlap heavily. Brain weight was about 58 g lower in dementia (p = 0.025), but this is mostly a sex effect: males had much heavier brains, and within each sex the dementia difference was not significant. Higher pTAU was linked to lower MMSE scores (R² = 0.086, p = 0.0087), but the relationship is modest and shows association, not cause.

**Recommendation.** A single tissue amyloid level should not be used alone to tell whether someone had dementia, brain weight comparisons must account for sex, and pTAU deserves further study in larger samples.

**Ethical implications.**
* **Biomarkers:** people can have high amyloid without ever developing symptoms, so treating a positive test as a dementia diagnosis could cause distress, stigma, and loss of independence.
* **Access:** lecanemab and donanemab require biomarker testing and frequent MRI monitoring. If these are available mostly at academic centers, rural and under-resourced patients could be left out.
* **Representation:** our donors come from one region and chose to join research studies, so cutoffs based on data like ours could misclassify under-represented groups.
* **Sex differences:** the brain weight gap is a normal baseline difference and should not be read as a difference in cognitive ability.

## Limitations and Future Work:

**Limitations.**
1. **Skewed data:** the amyloid values are extremely right-skewed (skewness around 7), which breaks the normality assumption of our tests. Our amyloid results are inconclusive, not proof of no difference.
2. **Small groups:** splitting 84 donors into four groups (dementia males n = 15) leaves little power to detect small or sex-specific effects.
3. **Multiple tests:** we ran many tests, and only Tukey's test corrects for this.
4. **MMSE timing:** cognition was measured a median of about 28 months before the tissue, and many scores sit near the top of the scale.
5. **Labels and coverage:** donors without dementia can still have Alzheimer's pathology, and all data come from one brain region and one region's cohort.
6. **Missing adjustments:** we did not adjust brain weight for age or head size, and we did not test whether pTAU and MMSE are related differently in males and females.
7. **Association only:** these are snapshots from donated brains and cannot show cause.

**Future work.**
* Repeat the protein analyses on log10-transformed values or with a Mann-Whitney U test.
* Use a two-way ANOVA or multiple regression to test sex and dementia together while controlling for age, post-mortem interval, education, and APOE.
* Fit the pTAU and MMSE regression separately for each sex.
* Use the dataset's Braak, Thal, and neuropathology scores, and repeat in a larger, more diverse cohort.
